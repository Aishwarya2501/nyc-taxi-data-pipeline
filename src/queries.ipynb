```python
%sql
select DATE_TRUNC('month', pickup_datetime) as month,
       cab_colour,
       count(*) as total_trips,
       sum(total_amount) as total_revenue
from taxi_trips
group by month,cab_colour
```

```python
%sql
with top_zones as (
select pickup_borough, pickup_zone,count(*) as total_trips ,
dense_rank() over (order by count(*) desc) as rn from taxi_trips 
group by 1,2)
select pickup_zone,pickup_borough,total_trips from top_zones where rn <= 10
order by total_trips desc
```

```python
%sql
select hour(pickup_datetime) as hour_of_the_day, 
round(avg(trip_distance),2) as avg_trip_distance, 
round(avg(fare_amount),2) as avg_fare_amount from taxi_trips
group by hour_of_the_day
order by hour_of_the_day
```
